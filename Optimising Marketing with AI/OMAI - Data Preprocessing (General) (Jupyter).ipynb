```python
import pandas as pd

def load_custom_review_file(path):
    reviews = []
    current_review = {}

    with open(path, 'r', encoding = 'utf-8') as f:
        for line in f:
            line = line.strip()

            if not line:
                if current_review:
                    reviews.append(current_review)
                    current_review = {}
                continue

            if ': ' in line:
                key, value = line.split(': ', 1)
                current_review[key.strip()] = value.strip()

        if current_review:
            reviews.append(current_review)

    return pd.DataFrame(reviews)

OMAI_reviews = load_custom_review_file("Video_Games.txt")

print(OMAI_reviews.head())
```

```python
import pandas as pd

def load_product_categories(path):
    data = []
    current_product_id = None
    current_categories = []

    with open(path, 'r', encoding = 'latin-1') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            if line.startswith('B0'):  
                if current_product_id is not None:
                    data.append({'productid': current_product_id,
                                 'categories': ', '.join(current_categories)})
                current_product_id = line
                current_categories = []
            else:
                current_categories.append(line)
        if current_product_id is not None:
            data.append({'productid': current_product_id,
                         'categories': ', '.join(current_categories)})

    return pd.DataFrame(data)

OMAI_categories = load_product_categories("categories.txt")
print(OMAI_categories.head())
```

```python
def override_gameboy(gbobservation):
    gb_variants = {'Game Boy', 'Game Boy Advance', 'Game Boy Color'}
    matched = set(gbobservation['consoles'])
    if matched.issubset(gb_variants):
        if matched == {'Game Boy', 'Game Boy Advance'}:
            return 'Game Boy Advance'
        elif matched == {'Game Boy', 'Game Boy Color'}:
            return 'Game Boy Color'
        elif matched == {'Game Boy', 'Game Boy Color', 'Game Boy Advance'}:
            return 'Game Boy'   
    return gbobservation['console']  
        
def override_pcmac (pcobservation):
    pc_variants = {'Mac', 'PC'}
    matched = set(pcobservation['consoles'])
    if matched.issubset(pc_variants):
        if matched == pc_variants:
            return 'PC & Mac'
    return pcobservation['console']  

def override_console(consoleobservation):
    consoleobservation['console'] = override_pcmac(consoleobservation)
    consoleobservation['console'] = override_gameboy(consoleobservation)
    return consoleobservation['console']

OMAI_reviews['console'] = OMAI_reviews.apply(override_console, axis = 1)
```
