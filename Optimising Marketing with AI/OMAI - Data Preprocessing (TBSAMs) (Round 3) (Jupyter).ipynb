### **Data Preprocessing - Transformer-Based Sentiment Analysis Models (Round 3)**

---

```python
import pandas as pd
pd.set_option('display.max_colwidth', None)

OMAI_reviews = pd.read_csv('OMAI - Data - SNAP Amazon Reviews (Video Games).csv')
OMAI_reviews.iloc[0:1]
```

```python
import numpy as np

def create_sentimentR3(rating):
    if rating == 1.0 or rating == 2.0 or rating == 3.0:
        return 0 
    elif rating == 4.0 or rating == 5.0:
        return 1 
OMAI_reviews['sentiment_scoreR3'] = OMAI_reviews['score'].apply(create_sentimentR3)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_scoreR3']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_labelR3(sentiment_scoreR3):
    if sentiment_scoreR3 == 1:
        return 'Positive' 
    elif sentiment_scoreR3 == 0:
        return 'Non-Positive'
OMAI_reviews['sentiment_score_labelR3'] = OMAI_reviews['sentiment_scoreR3'].apply(sentiment_score_labelR3)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_scoreR3', 
                                   'sentiment_score_labelR3']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
vader = SentimentIntensityAnalyzer()

OMAI_reviews['combined_text'] = OMAI_reviews['summary'].fillna('') + ' ' + OMAI_reviews['review'].fillna('')
OMAI_reviews['sentiment_vaderR3'] = OMAI_reviews['combined_text'].apply(lambda x: vader.polarity_scores(x))
OMAI_reviews['sentiment_vader_compoundR3'] = OMAI_reviews['sentiment_vaderR3'].apply(lambda x: x['compound'])
OMAI_reviews['sentiment_vader_posR3'] = OMAI_reviews['sentiment_vaderR3'].apply(lambda x: x['pos'])
OMAI_reviews['sentiment_vader_neuR3'] = OMAI_reviews['sentiment_vaderR3'].apply(lambda x: x['neu'])
OMAI_reviews['sentiment_vader_negR3'] = OMAI_reviews['sentiment_vaderR3'].apply(lambda x: x['neg'])
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_scoreR3', 
                                   'sentiment_score_labelR3', 'sentiment_vader_compoundR3',
                                   'sentiment_vader_posR3', 'sentiment_vader_neuR3', 
                                   'sentiment_vader_negR3']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_vader_R3(sentiment_vader_compoundR3):
    if sentiment_vader_compoundR3 >= 0.05:
        return 1 
    else:
        return 0 
OMAI_reviews['sentiment_score_vader_R3'] = OMAI_reviews['sentiment_vader_compoundR3'].apply(sentiment_score_vader_R3)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score',
                                   'sentiment_vader_compoundR3', 
                                   'sentiment_score_vader_R3']].drop_duplicates()
OMAI_reviews_check[10000:10010]
```

```python
import numpy as np

def sentiment_score_vader_labelR3(sentiment_score_vader_R3):
    if sentiment_score_vader_R3 == 1:
        return 'Positive' 
    elif sentiment_score_vader_R3 == 0:
        return 'Non-Positive'
OMAI_reviews['sentiment_score_vader_labelR3'] = OMAI_reviews['sentiment_score_vader_R3'].apply(sentiment_score_vader_labelR3)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score_vader_R3', 
                                   'sentiment_score_vader_labelR3']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```
