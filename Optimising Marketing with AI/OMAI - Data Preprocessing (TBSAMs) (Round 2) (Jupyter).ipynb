### **Data Preprocessing - Transformer-Based Sentiment Analysis Models (Round 2)**

---

```python
import pandas as pd
pd.set_option('display.max_colwidth', None)

OMAI_reviews = pd.read_csv('OMAI - Data - SNAP Amazon Reviews (Video Games).csv')
OMAI_reviews.iloc[0:1]
```

```python
import numpy as np

def create_sentiment(rating):
    if rating == 1.0 or rating == 2.0:
        return -1 
    elif rating == 4.0 or rating == 5.0:
        return 1 
    else:
        return 0 
OMAI_reviews['sentiment_score'] = OMAI_reviews['score'].apply(create_sentiment)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_label(sentiment_score):
    if sentiment_score == 1:
        return 'Positive' 
    elif sentiment_score == -1:
        return 'Negative'
    else:
        return 'Neutral'
OMAI_reviews['sentiment_score_label'] = OMAI_reviews['sentiment_score'].apply(sentiment_score_label)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score', 
                                   'sentiment_score_label']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
vader = SentimentIntensityAnalyzer()

OMAI_reviews['combined_text'] = OMAI_reviews['summary'].fillna('') + ' ' + OMAI_reviews['review'].fillna('')
OMAI_reviews['sentiment_vaderR2'] = OMAI_reviews['combined_text'].apply(lambda x: vader.polarity_scores(x))
OMAI_reviews['sentiment_vader_compoundR2'] = OMAI_reviews['sentiment_vaderR2'].apply(lambda x: x['compound'])
OMAI_reviews['sentiment_vader_posR2'] = OMAI_reviews['sentiment_vaderR2'].apply(lambda x: x['pos'])
OMAI_reviews['sentiment_vader_neuR2'] = OMAI_reviews['sentiment_vaderR2'].apply(lambda x: x['neu'])
OMAI_reviews['sentiment_vader_negR2'] = OMAI_reviews['sentiment_vaderR2'].apply(lambda x: x['neg'])
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score', 'sentiment_score_label', 
                                   'sentiment_vader_compoundR2', 'sentiment_vader_posR2', 'sentiment_vader_neuR2', 
                                   'sentiment_vader_negR2']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_nn(sentiment_score):
    if sentiment_score == 1:
        return 2 
    elif sentiment_score == 0:
        return 1
    else:
        return 0 
OMAI_reviews['sentiment_score_nn'] = OMAI_reviews['sentiment_score'].apply(sentiment_score_nn)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score', 
                                   'sentiment_score_nn']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_vader_nnR2(sentiment_vader_compoundR2):
    if sentiment_vader_compoundR2 >= 0.05:
        return 2 
    elif sentiment_vader_compoundR2 <= -0.05:
        return 0
    else:
        return 1 
OMAI_reviews['sentiment_score_vader_nnR2'] = OMAI_reviews['sentiment_vader_compoundR2'].apply(sentiment_score_vader_nnR2)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 
                                   'sentiment_vader_compoundR2', 
                                   'sentiment_score_vader_nnR2']].drop_duplicates()
OMAI_reviews_check[10000:10010]
```
