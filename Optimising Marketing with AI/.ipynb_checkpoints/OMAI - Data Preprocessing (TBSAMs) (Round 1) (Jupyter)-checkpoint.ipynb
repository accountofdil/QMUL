### **Data Preprocessing - Transformer-Based Sentiment Analysis Models (Round 1)**

---

```python
import pandas as pd
pd.set_option('display.max_colwidth', None)

OMAI_reviews = pd.read_csv('OMAI - Data - SNAP Amazon Reviews (Video Games).csv')
OMAI_reviews.iloc[0:1]
```

```python
OMAI_reviews['description'] = OMAI_reviews['description'].str.lower()
OMAI_reviews['summary'] = OMAI_reviews['summary'].str.lower()
OMAI_reviews['review'] = OMAI_reviews['review'].str.lower()
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'summary', 'review', 'description']].drop_duplicates()
OMAI_reviews_check.iloc[90:93]
```

```python
import numpy as np

def create_sentiment(rating):
    if rating == 1.0 or rating == 2.0:
        return -1 
    elif rating == 4.0 or rating == 5.0:
        return 1 
    else:
        return 0 
OMAI_reviews['sentiment_score'] = OMAI_reviews['score'].apply(create_sentiment)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_label(sentiment_score):
    if sentiment_score == 1:
        return 'Positive' 
    elif sentiment_score == -1:
        return 'Negative'
    else:
        return 'Neutral'
OMAI_reviews['sentiment_score_label'] = OMAI_reviews['sentiment_score'].apply(sentiment_score_label)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score', 'sentiment_score_label']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
vader = SentimentIntensityAnalyzer()

OMAI_reviews['combined_text'] = OMAI_reviews['summary'].fillna('') + ' ' + OMAI_reviews['review'].fillna('')
OMAI_reviews['sentiment_vader'] = OMAI_reviews['combined_text'].apply(lambda x: vader.polarity_scores(x))
OMAI_reviews['sentiment_vader_compound'] = OMAI_reviews['sentiment_vader'].apply(lambda x: x['compound'])
OMAI_reviews['sentiment_vader_pos'] = OMAI_reviews['sentiment_vader'].apply(lambda x: x['pos'])
OMAI_reviews['sentiment_vader_neu'] = OMAI_reviews['sentiment_vader'].apply(lambda x: x['neu'])
OMAI_reviews['sentiment_vader_neg'] = OMAI_reviews['sentiment_vader'].apply(lambda x: x['neg'])
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 'sentiment_score', 'sentiment_score_label', 
                                   'sentiment_vader_compound', 'sentiment_vader_pos', 'sentiment_vader_neu', 
                                   'sentiment_vader_neg']].drop_duplicates()
OMAI_reviews_check[200000:200005]
```

```python
import numpy as np

def sentiment_score_nn(sentiment_score):
    if sentiment_score == 1:
        return 2 
    elif sentiment_score == 0:
        return 1
    else:
        return 0 
OMAI_reviews['sentiment_score_nn'] = OMAI_reviews['sentiment_score'].apply(sentiment_score_nn)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 
                                   'sentiment_score', 'sentiment_score_nn']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
import numpy as np

def sentiment_score_vader_nn(sentiment_vader_compound):
    if sentiment_vader_compound >= 0.05:
        return 2 
    elif sentiment_vader_compound <= -0.05:
        return 0
    else:
        return 1 
OMAI_reviews['sentiment_score_vader_nn'] = OMAI_reviews['sentiment_vader_compound'].apply(sentiment_score_vader_nn)
```

```python
OMAI_reviews_check = OMAI_reviews[['productid', 'score', 
                                   'sentiment_vader_compound', 'sentiment_score_vader_nn']].drop_duplicates()
OMAI_reviews_check[20000:20010]
```

```python
OMAI_reviews = OMAI_reviews.copy(deep = True)
OMAI_reviews.to_csv('OMAI - Data (TBSAMs) (Round 1).csv')
```
