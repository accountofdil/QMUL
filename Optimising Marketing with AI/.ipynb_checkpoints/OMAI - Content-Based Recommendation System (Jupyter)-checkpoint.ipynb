### **Content-Based Recommendation System**

---

```python
import pandas as pd
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

pd.set_option('display.max_columns', None)

OMAI_mlrm_logregknn = pd.read_csv('OMAI - Results - MLRMs (LR & K-Nearest Neighbours) (Predictions).csv')
OMAI_mlrm_logregknn = OMAI_mlrm_logregknn.drop('Unnamed: 0', axis = 1)
OMAI_mlrm_logregknn.iloc[0:3]
```

```python
logreg_threshold = 0.55
OMAI_mlrm_logregknn['logreg_thresholds'] = np.where(OMAI_mlrm_logregknn['logreg_prob'].values >= logreg_threshold,
                                                  'Recommend', 'Not Recommend')
OMAI_mlrm_logregknn_reduced = (OMAI_mlrm_logregknn.loc[OMAI_mlrm_logregknn['logreg_thresholds'] == 'Recommend']
                             .sort_values(by = 'logreg_prob', ascending = False).copy())
```

```python
tfidf_rec = TfidfVectorizer(stop_words = 'english')
OMAI_mlrm_logregknn['description'] = OMAI_mlrm_logregknn['description'].fillna('')
tfidf_matrix_rec = tfidf_rec.fit_transform(OMAI_mlrm_logregknn['description'])
tfidf_matrix_rec.shape
tfidf_array_rec = tfidf_matrix_rec.toarray()
print(tfidf_array_rec)
cosine_similarity = linear_kernel(tfidf_matrix_rec, tfidf_matrix_rec)
cosine_similarity.shape
cosine_similarity[1]
```

```python
indices = pd.Series(OMAI_mlrm_logregknn.index, index = OMAI_mlrm_logregknn['title']).drop_duplicates()

def recommend_games(title, 
                    df_reduced = OMAI_mlrm_logregknn_reduced, 
                    cosine_similarity_matrix = cosine_similarity, 
                    top10recommendations = 10):
    
    idx = indices[title]  
    similarity_scores = list(enumerate(cosine_similarity_matrix[idx, :].ravel()))
    similarity_scores = sorted(similarity_scores, key = lambda x: x[1], reverse = True)
    similarity_scores = similarity_scores[1:]  
    position_to_score = dict(similarity_scores)

    positions = set(df_reduced.index)
    filtered_positions = [i for i, _ in similarity_scores if i in positions]

    return df_reduced.loc[filtered_positions, ['title', 'console', 'price', 'logreg_prob']] \
                     .assign(similarity_score = [position_to_score[i] for i in filtered_positions]) \
                     .head(top10recommendations)
```
